```python
import pandas as pd
import numpy as np

np.random.seed(0)

n = 1500

# Creation of the DataFrame
df = pd.DataFrame({
    "age_group" : np.random.choice(
        ["18-34", "35-49", "50-64", "65+"],
        size = n,
        p=[0.3, 0.25, 0.25, 0.2]
    ),
    "sex": np.random.choice(["Female", "Male"], size=n),
    "insurance": np.random.choice(
        ["Private", "Public", "Uninsured"], 
        size=n, 
        p=[0.5, 0.35, 0.15]
    ),
    "preventive_visit": np.random.choice(
        ["Yes", "No"], 
        size=n, 
        p=[0.55, 0.45]
    )
})

# First 5 rows
df.head()
```
